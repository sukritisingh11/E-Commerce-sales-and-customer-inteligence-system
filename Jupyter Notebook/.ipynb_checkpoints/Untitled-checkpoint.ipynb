# Create the sales dataset

```python
sales = orders.merge(
    items,
    on="order_id",
    how="inner"
)

print("Sales dataset shape:", sales.shape)
sales.head()
```

```python
sales = sales.merge(
    products,
    on="product_id",
    how="left"
)

print("After merging products:", sales.shape)
sales.head()
```

```python
sales["revenue"] = sales["price"] * sales["order_item_id"].apply(
    lambda x: 1
)
```

```python
sales["revenue"] = sales["price"]
```

```python
sales[["order_id", "product_id", "price", "revenue"]].head()

```

```python
total_revenue = sales["revenue"].sum()

print("Total Revenue: ₹", round(total_revenue, 2))

```

```python
top_products = (
    sales.groupby("product_category_name")["revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products

```

```python
plt.figure(figsize=(10,6))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Product Categories by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product Category")

plt.show()
```

```python
sales["month"] = sales["order_purchase_timestamp"].dt.to_period("M")

```

```python
monthly_sales = (
    sales.groupby("month")["revenue"]
    .sum()
)

monthly_sales

```

# — Customer Intelligence Dataset

```python
customer_sales = sales.merge(
    customers,
    on="customer_id",
    how="left"
)

customer_sales.head()
```

```python
reference_date = sales["order_purchase_timestamp"].max()

print("Reference Date:", reference_date)

```

```python
rfm = customer_sales.groupby("customer_unique_id").agg(
    recency=("order_purchase_timestamp",
             lambda x: (reference_date - x.max()).days),

    frequency=("order_id", "nunique"),

    monetary=("revenue", "sum")
).reset_index()

rfm.head()
```

```python
rfm.shape
```

 first ML model: Customer Segmentation using K-Means

```python
rfm.head()
```

```python
rfm[["recency", "frequency", "monetary"]].describe()

```

```python
fm_ml = rfm.copy()

rfm_ml["recency"] = np.log1p(rfm_ml["recency"])
rfm_ml["frequency"] = np.log1p(rfm_ml["frequency"])
rfm_ml["monetary"] = np.log1p(rfm_ml["monetary"])
```

```python
rfm_ml = rfm.copy()

rfm_ml["recency"] = np.log1p(rfm_ml["recency"])
rfm_ml["frequency"] = np.log1p(rfm_ml["frequency"])
rfm_ml["monetary"] = np.log1p(rfm_ml["monetary"])
```

```python
rfm_ml.head()

```

```python
from sklearn.preprocessing import StandardScaler
```

```python
features = [
    "recency",
    "frequency",
    "monetary"
]

scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(
    rfm_ml[features]
)

print(rfm_scaled[:5])
```

```python
from sklearn.cluster import KMeans

inertia = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(rfm_scaled)

    inertia.append(kmeans.inertia_)

```

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(2, 11),
    inertia,
    marker="o"
)

plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method for Customer Segmentation")

plt.show()
```

Train K-Means

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

rfm["cluster"] = kmeans.fit_predict(rfm_scaled)

rfm.head()
```

```python
rfm["cluster"].value_counts().sort_index()

```

```python
cluster_summary = rfm.groupby("cluster").agg(
    customers=("customer_unique_id", "count"),
    avg_recency=("recency", "mean"),
    avg_frequency=("frequency", "mean"),
    avg_monetary=("monetary", "mean")
).round(2)

cluster_summary
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=rfm,
    x="frequency",
    y="monetary",
    hue="cluster",
    palette="viridis"
)

plt.title("Customer Segmentation")
plt.xlabel("Purchase Frequency")
plt.ylabel("Customer Spending")

plt.show()
```

```python
rfm.to_csv(
    "customer_segmentation.csv",
    index=False
)

print("Customer segmentation dataset saved!")
```
